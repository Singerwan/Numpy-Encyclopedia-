## squeeze vs expanddims

In NumPy, np.squeeze removes single-dimensional entries from an array’s shape,
<br>while np.expanddims adds a new axis at a specified position. They are essentially inverse operations.

```python
import numpy as np
# Original 1D array
arr = np.array([1, 2, 3])
print("Original:", arr, arr.shape) # (3,)
# Expand dimensions at axis=0
expanded = np.expand_dims(arr, axis=0)
print("Expanded:", expanded, expanded.shape) # (1, 3)
# Squeeze back to original shape
squeezed = np.squeeze(expanded)
print("Squeezed:", squeezed, squeezed.shape) # (3,)
```

##### argsort

The np.argsort(x) function in NumPy returns the indices that would sort the array x. It performs an indirect sort, meaning it does not modify the original array but provides the indices to rearrange it in sorted order.

```python
import numpy as np
x = np.array([3, 1, 4, 1, 5])
indices = np.argsort(x)
print("Original array:", x) # [3, 1, 4, 1, 5]
print("Indices to sort:", indices) # [1, 3, 0, 2, 4]
print("Sorted array:", x[indices]) # [1, 1, 3, 4, 5]
```
